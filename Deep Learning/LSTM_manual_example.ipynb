#Forget gate

```python
w_f = np.array([
    [.1, -.2,.7 ,.8,.7,.9, .1],
    [.2,-.5,.1,.6,.8,-.9,.3],
    [.6,.1,.3,-.4,-.3,-.1,-.56]


])
```

```python
w_f
```

```python
h_x  = np.array([[
    .1,
    .2,
    .7,
    0,
    1,
    0,
    0
]]).T
```

```python
h_x
```

```python
b_f = np.array([
       [-0.3],
       [0.2],
       [0.4],
                ])
b_f
```

```python
def sigmoid(x):
  result = (1/(1+np.exp(-x)))
  result[result > 0.5] = 1
  result[result<=0.5] =0
  return result
```

```python
f_t =  sigmoid(  np.dot(w_f, h_x) + b_f  )
f_t
```

# Candidate cell

```python
w_c = w_f = np.array([
    [.41, .32,-.7 ,.8,.7,.9, .1],
    [.22,-.25,.41,.6,-.18,-.39,-.13],
    [-.16,.31,-.13,-.54,-.73,.51,-.56]


])
```

```python
b_c = np.array([
       [-0.9],
       [-0.7],
       [0.4],
                ])
b_c
```

```python
c_tilda = np.tanh(  np.dot( w_c , h_x ) + b_c  )

c_tilda
```

```python
np.dot(w_f, h_x) + b_f
```

#Input gate

```python
w_i = np.array([
    [-.41, .32,-.27 ,.98,.17,.39, .51],
    [.22,-.25,.541,.66,-.318,-.329,-.13],
    [-.916,.831,-.613,-.54,-.273,.51,-.56]


])
b_i = np.array([
       [-0.9],
       [0.7],
       [0.4],
                ])
```

```python
i_t = sigmoid(  np.dot(w_i, h_x) + b_i  )
i_t
```

```python
c_t_old = np.array([
       [.1],
       [.9],
       [.5]])
```

#C_t update

```python
c_t = (f_t*c_t_old) + (i_t * c_tilda)
```

```python
c_t
```

# Output gate

```python
w_o = np.array([
    [-.1, .32,.27 ,.8,.17,.39, .51],
    [.22,-.25,.541,.66,-.38,-.39,.13],
    [-.96,.831,.613,-.94,-.73,.51,-.56]


])
b_o = np.array([
       [-0.3],
       [0.4],
       [0.1],
                ])
```

```python
o_t = sigmoid(  np.dot(w_o, h_x) + b_o  )
o_t
```

#h_t update

```python
h_t = o_t* np.tanh(c_t)
h_t
```

#Predicting next word

```python
w_y = np.array([
    [0.2, -.9,-.8],
    [-.1, .2, .3],
    [.3,.4,.7],
    [-.3,.4,.7]

])
w_y
```

```python
b_y = np.array([[1.],
       [1.],
       [0.],
        [.5]])
```

```python
def softmax(x):
  result = []
  for i in x:
    result.append(np.exp(i)/np.sum(np.exp(x))   )
  return(np.array(result))
```

```python
y = softmax(np.dot(w_y, h_t) + b_y)
y
```
